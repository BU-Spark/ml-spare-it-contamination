```python
from rembg import remove
from PIL import Image
import numpy as np
import io
import os
def remove_background(input_image_path, output_image_path, mask_output_path):
    """
    Removes the background from an image and creates a mask for the remaining foreground.

    Args:
        input_image_path: Path to the input image file.
        output_image_path: Path to save the image with the background removed.
        mask_output_path: Path to save the generated mask image.
    """
    try:
        base_name = os.path.splitext(os.path.basename(input_image_path))[0]
        output_image_path = os.path.join(output_image_path, f"{base_name}_background_removed.png")
        mask_output_path = os.path.join(mask_output_path, f"{base_name}_mask.png")

        with open(input_image_path, "rb") as img_file:
            input_image = img_file.read()

        output_image_data = remove(input_image)

        with open(output_image_path, "wb") as out_file:
            out_file.write(output_image_data)

        result_image = Image.open(io.BytesIO(output_image_data)).convert("RGBA")
        mask = result_image.split()[-1]
        mask.save(mask_output_path, "PNG")

        print(f"Background removed and images saved: {output_image_path}, {mask_output_path}")

    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
#Use eample
input_image_path = "/content/generated_image_2.png"
output_image_path = "/content/output"
mask_output_path = "/content/output"

remove_background(input_image_path, output_image_path, mask_output_path)
```
